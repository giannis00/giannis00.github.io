# 🚀 From One Table to a Cluster: Working with Data at Scale

### A complete, hands-on course: run entirely in your browser
**Instructor:** Ioannis Tsioulis

---

Welcome. This course is about what happens to data once there is a lot of it. It starts with a single table that you can read on one screen, and ends with computations spread over many machines and data that never stops arriving. Every notebook opens **directly in your browser** with Google Colab: there is nothing to install and nothing to download.

Each idea is first explained in plain words, then built **by hand** in a few lines of Python, and only then handed over to a professional tool. You will write a join, an index lookup, a document store, a MapReduce job and a streaming window yourself before using the systems that do these things for a living. A database or a cluster stops being a black box once you have built a small one.

## Who this course is for

You should be comfortable with Python: loops, functions, lists and dictionaries. No knowledge of databases is assumed.

This course does **not** re-teach what its companion courses cover. Loading and exploring a table with pandas, plotting, and machine learning are in *From Python to Deep Learning & Fine-Tuning*. Big O notation, binary search, hash tables and trees are in *From Zero to Data Structures & Algorithms in Python*. Where a notebook here relies on one of those ideas, it says so and points to the exact notebook. Everything here is about **storing, querying and processing data as it grows**.

## How to use these notebooks

- Work through them **in order**, each one builds on the last.
- **Run every code cell yourself** (click it and press `Shift + Enter`). Then change a number, such as the size of the data, and run it again. Scale is something you have to feel.
- Attempt every exercise **before** opening the solution.
- Every notebook creates its own data, so nothing has to be downloaded.
- Use **File → Save a copy in Drive** to keep your own editable version of each notebook.

## A note on the learning philosophy

There is a single question running through the whole course: **"what changes when the data no longer fits?"** First it no longer fits in one table, and we need joins. Then a query no longer finishes in reasonable time, and we need indexes. Then the data no longer fits one shape, then one computer's memory, then one computer at all, and finally it no longer fits in time, because it keeps arriving. Each module is the answer to one of those moments.

We also never take a claim on faith. When we say that one approach is faster or smaller than another, we **measure it**, and we look just as carefully for the cases where the simple approach wins. A large part of working with data well is knowing when you do **not** need the heavy machinery.
