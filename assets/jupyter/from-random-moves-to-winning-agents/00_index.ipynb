Welcome. This course is about programs that learn the way people learn to ride a bicycle: by trying, failing, and gradually keeping what works. Nobody shows them the right answer. They act, they receive a reward, and they have to work out for themselves how to get more of it. That is **reinforcement learning**.

Every idea is taught with a **game**. The course starts with a row of slot machines and ends with agents that land a spacecraft, play from the pixels of the screen, think ahead in Connect Four and train against themselves. Every notebook opens **directly in your browser** with Google Colab, and in almost every one you will **watch a video** of the agent you have just trained.

Each method is first explained in plain words, then written **by hand** in a few lines of Python, and only then taken from a library. A Q-table, a replay buffer or a policy gradient stops being a black box once you have built a small one yourself.

## Who this course is for

You should be comfortable with Python: functions, loops, lists, dictionaries and classes, and a little NumPy. Modules A and B need nothing more.

From Module C onwards the agents are neural networks written in PyTorch. This course does **not** re-teach neural networks. They are covered in the companion course *From Python to Deep Learning & Fine-Tuning* (Notebooks 12 to 14), and wherever a notebook here relies on an idea from there, it says so and points to the notebook.

No GPU is needed for anything that runs by default.

## How to use these notebooks

- Work through them **in order**. Each one builds on the last, and later notebooks refer to experiments in earlier ones.
- **Run every code cell yourself** (click it and press `Shift + Enter`), then change something and run it again: the reward, the amount of exploration, the random seed.
- Attempt every exercise **before** opening the solution.
- **Training takes time.** Most cells finish in seconds. Some, which train an agent on a harder game, take a few minutes on Colab's free machines, and the notebook always tells you beforehand.
- **Your numbers will differ from ours.** Reinforcement learning is noisy, and two runs of the same program can end in different places. The course treats that as a subject in its own right: results are compared over several runs, and the text says so when one run is not enough.
- Use **File → Save a copy in Drive** to keep your own editable version of each notebook.

## A note on the learning philosophy

One question runs through the whole course: **"how can an agent find good behaviour when nobody tells it what good behaviour is?"** The first answer, in Notebook 01, is a table of averages and a coin flip for exploration. It is surprisingly useful and obviously limited. Each module after that keeps what worked and repairs what did not: delayed rewards, unknown rules, states too many to list, actions that are numbers, opponents that learn.

We also never take a claim on faith. When we say that one method is better than another, we **measure it**, and when a measurement does not support the claim, the notebook says so. Several of the experiments in this course end with "no clear difference", and that is as much a result as any other.
