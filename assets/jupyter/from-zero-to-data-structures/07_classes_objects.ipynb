## 1. From separate variables to objects

### The motivation
Suppose we want to store information about a person. So far we would use separate variables, or a dictionary:

```python
person = {"name": "Alice", "age": 30}
```

This holds the **data**, but the code that works on it (greeting the person, celebrating a birthday) lives somewhere else, in separate functions. Nothing ties the two together.

### Intuition first
A **class** bundles the data and the functions that work on that data into one unit.

- A **class** is a **blueprint**: the plan of a house.
- An **object** is one concrete thing built from the blueprint: an actual house.

From one plan you can build many houses. They share the same structure, but each has its own colour, its own address, its own residents.

### Terminology

| Term | Meaning |
|---|---|
| **class** | the blueprint |
| **object** (or *instance*) | one concrete thing built from the class |
| **attribute** | a variable that belongs to an object (its data) |
| **method** | a function that belongs to an object (its behaviour) |

## 2. Your first class

### Anatomy of a class
```python
class Person:
    def __init__(self, name, age):     # runs automatically when an object is created
        self.name = name               # store the data inside the object
        self.age = age
```
- By convention class names start with a **capital letter**.
- `__init__` (two underscores on each side) is the **initialiser**. Python calls it automatically every time a new object is created.
- `self` is **the object itself**. `self.name = name` means *"store the value of `name` inside this particular object."*

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Creating an object: call the class like a function
person1 = Person("Alice", 30)

# Reading its attributes with a dot
print(person1.name)
print(person1.age)
````

Notice that `__init__` has three parameters, yet we passed only two arguments. Python fills in `self` automatically with the object being created.

## 3. Adding methods

A **method** is a function defined inside a class. Its first parameter is always `self`, which gives it access to the attributes of the object it was called on.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

person1 = Person("Alice", 30)
person1.greet()          # the method is called with a dot, like an attribute
````

### Each object is independent
Every object has its **own copy** of the attributes. Changing one object does not affect another.

````python
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)
person3 = Person("Carol", 41)

person1.greet()
person2.greet()
person3.greet()
````

## 4. Changing the state of an object

The values of an object's attributes at a given moment are called its **state**. Attributes can be changed directly from outside. Better still, a method can change them, which keeps the logic in one place.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

    def have_birthday(self):
        self.age += 1                       # the method changes the object
        print(f"Happy birthday, {self.name}! You are now {self.age}.")

person1 = Person("Alice", 30)
person1.greet()

person1.age = 31            # changing an attribute directly
person1.greet()

person1.have_birthday()     # changing it through a method
person1.greet()
````

### A worked example: a `Car` class
A method can take extra parameters after `self`, and it can return a value like any function.

````python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.mileage = 0                 # every new car starts at zero

    def drive(self, km):
        self.mileage += km

    def description(self):
        return f"{self.year} {self.make} {self.model}, {self.mileage} km"

car1 = Car("Toyota", "Corolla", 2020)
car2 = Car("Fiat", "Panda", 2015)

car1.drive(120)
car1.drive(80)
car2.drive(15)

print(car1.description())
print(car2.description())
````

> ⚠️ **Common pitfalls**
>
> - Forgetting `self` as the first parameter of a method: `def greet():` fails with a `TypeError` when called.
> - Forgetting `self.` when using an attribute: inside a method, `name` is an unknown variable, `self.name` is the attribute.
> - Writing `_init_` with one underscore on each side. The initialiser needs **two**: `__init__`.

## 5. Protecting the state of an object

### Why it matters
A method is the right place to put **rules**. A bank account must never allow a withdrawal larger than its balance. If every change to the balance goes through a method, the rule lives in exactly one place and can never be forgotten.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("The amount must be positive.")
            return
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
            return
        self.balance -= amount
        print(f"Withdrew {amount}. New balance: {self.balance}")

account = BankAccount("Anna", 100)
account.deposit(50)
account.withdraw(500)       # refused by the rule inside the method
account.withdraw(30)
````

## 6. Controlling how an object is displayed: `__str__`

Printing an object normally shows something unhelpful, like `<__main__.Person object at 0x7f...>`. If a class defines the special method `__str__`, Python uses the string it returns whenever the object is printed.

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"'{self.title}' by {self.author} ({self.pages} pages)"

book = Book("Dune", "Frank Herbert", 612)
print(book)
````

## 7. Objects working together

Attributes can hold anything, including **lists of other objects**. This is how larger programs are organised: each class does one job, and objects cooperate.

````python
class Library:
    def __init__(self):
        self.books = []                  # starts empty

    def add_book(self, book):
        self.books.append(book)

    def total_pages(self):
        total = 0
        for book in self.books:
            total += book.pages
        return total

    def show(self):
        for book in self.books:
            print("-", book)             # uses Book.__str__

library = Library()
library.add_book(Book("Dune", "Frank Herbert", 612))
library.add_book(Book("Emma", "Jane Austen", 474))

library.show()
print("Total pages:", library.total_pages())
````

> 🧠 Look closely at `Library`. It **wraps a list** and offers a small set of methods to work with it. That is exactly the shape of the stacks, queues and other data structures we will build in Module D.

## 8. Inheritance: building on an existing class

### Intuition first
Sometimes a new class is *a special kind of* an existing one. A teacher **is a** person, with everything a person has, plus a subject to teach. Rewriting the whole `Person` class would be wasteful. **Inheritance** lets the new class start with everything the old one has, and add or change only what is different.

### Terminology
- The existing class is the **parent** (or *base*) class.
- The new class is the **child** (or *derived*) class. It is written `class Child(Parent):`.
- `super().__init__(...)` calls the initialiser of the parent, so that we do not repeat its code.
- A child may **override** a method by defining one with the same name.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name}.")

class Teacher(Person):                    # Teacher inherits from Person
    def __init__(self, name, age, subject):
        super().__init__(name, age)       # let Person set up name and age
        self.subject = subject            # then add what is new

    def teach(self):                      # a new method
        print(f"{self.name} teaches {self.subject}.")

teacher = Teacher("Eleni", 42, "Mathematics")
teacher.greet()          # inherited from Person, we did not write it again
teacher.teach()          # defined in Teacher
````

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def __str__(self):
        return f"{self.name}: average {self.average():.1f}"

student = Student("Eleni")
student.add_grade(8)
student.add_grade(9)
student.add_grade(6)
print(student)
```

*The check for an empty list is an **edge case**: without it, `average()` on a new student would divide by zero.*

```python
class Playlist:

    def add_song(self, song):
```

## 🔑 Key takeaways

- A **class** is a blueprint; an **object** is a concrete instance built from it.
- `__init__` runs automatically when an object is created and sets up its **attributes**.
- `self` is the object itself; attributes are always written `self.something` inside the class.
- **Methods** are functions that belong to an object; they can read and change its **state**.
- Each object has its own independent copy of the attributes.
- Put the **rules** inside methods, so they live in one place.
- `__str__` controls how an object is printed; objects can contain lists of other objects.
- **Inheritance** (`class Child(Parent):`) lets a new class reuse an existing one and add or override only what differs.
