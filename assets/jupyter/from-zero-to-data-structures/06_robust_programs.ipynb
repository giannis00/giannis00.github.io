## 1. Errors are information

### Intuition first
Beginners often feel that an error message means they have failed. Experienced programmers see dozens every day and treat them as what they are: **the computer telling you precisely what went wrong and where**.

### How to read an error
Always read the **last line first**. It contains the type of the error and a short explanation. Then look at the line number above it.

| Error | What it usually means |
|---|---|
| `SyntaxError` | the code breaks the grammar of Python: a missing colon, bracket or quote |
| `IndentationError` | the indentation is inconsistent |
| `NameError` | a name that does not exist: usually a typing mistake |
| `TypeError` | an operation on the wrong type, such as `"5" + 1` |
| `ValueError` | the right type but an impossible value, such as `int("hello")` |
| `ZeroDivisionError` | a division by zero |
| `IndexError` | a list index that is out of range |
| `KeyError` | a dictionary key that does not exist |

The first two are found before the program runs. All the others happen **while it runs**, and they are called **exceptions**.

## 2. Validating input with a loop

### Intuition first
A program that asks for an age and receives `-5` should not carry on. The rule is simple: **never trust input**. Check it, and if it is not acceptable, ask again.

### The pattern
Ask once, then use a `while` loop whose condition describes a **bad** answer. The loop keeps asking for as long as the answer is bad.

````python
# Divisibility checker: both numbers must be positive
a = int(input("Enter the dividend (a): "))
while a <= 0:
    print("a must be a positive integer.")
    a = int(input("Enter the dividend (a): "))

b = int(input("Enter the divisor (b): "))
while b <= 0:
    print("b must be a positive integer.")
    b = int(input("Enter the divisor (b): "))

if a % b == 0:
    print(f"{a} is divisible by {b}.")
else:
    print(f"{a} is not divisible by {b}. The remainder is {a % b}.")
````

The same validation was written twice, once for each number. Whenever you copy and paste code, that is a signal to write a **function**.

````python
def read_positive_int(prompt):
    number = int(input(prompt))
    while number <= 0:
        print("Please enter a positive integer.")
        number = int(input(prompt))
    return number

n = read_positive_int("How many students? ")
print(f"Registered {n} students.")
````

## 3. Catching errors with `try` / `except`

### Why it matters
The function above still has a weakness. If the user types `seven` instead of `7`, `int("seven")` raises a `ValueError` and the whole program stops. No condition can prevent this, because the failure happens *during* the conversion.

### Formal definition
```python
try:
    risky code            # Python attempts this
except SomeError:
    recovery code         # runs only if that error happened
```
If everything in the `try` block succeeds, the `except` block is skipped. If the named error occurs, Python jumps straight to the `except` block and the program continues.

````python
try:
    number = int(input("Enter a whole number: "))
    print(f"Twice your number is {number * 2}.")
except ValueError:
    print("That was not a whole number.")

print("The program is still running.")
````

### Combining `try` with a loop
Putting the `try` inside a `while True` loop gives the most robust input function you can write: it keeps asking until the input is both convertible **and** acceptable.

````python
def read_positive_int(prompt):
    while True:
        try:
            number = int(input(prompt))
            if number > 0:
                return number            # good input: leave the function
            print("The number must be positive.")
        except ValueError:
            print("Please type digits only.")

age = read_positive_int("Enter your age: ")
print(f"Your age is {age}.")
````

### A worked example: a simple calculator
Two different things can go wrong here: the user may type something that is not a number, and the user may ask to divide by zero. We handle each one.

````python
try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))
    operation = input("Choose an operation (+, -, *, /): ")

    if operation == "+":
        print("Result:", num1 + num2)
    elif operation == "-":
        print("Result:", num1 - num2)
    elif operation == "*":
        print("Result:", num1 * num2)
    elif operation == "/":
        print("Result:", num1 / num2)
    else:
        print("Unknown operation.")
except ValueError:
    print("Invalid input: please enter numbers.")
except ZeroDivisionError:
    print("Error: division by zero is not allowed.")
````

> ⚠️ **Common pitfalls**
>
> - Writing a bare `except:` with no error name. It hides **every** problem, including your own typing mistakes. Always name the error you expect.
> - Wrapping the whole program in one giant `try`. Keep the `try` block small, around the lines that can actually fail.

## 4. Solving a problem step by step

### Intuition first
Faced with a larger task, beginners try to write the whole program at once, and then cannot tell which part is wrong. Professionals do the opposite:

1. **Understand** the task. State the input and the expected output. Work one example by hand.
2. **Break it down** into small steps written in plain language.
3. **Write one step**, run it, and check it.
4. **Add the next step.** Repeat.
5. **Test** with ordinary values, with edge cases, and with wrong input.

### A worked example: a password validator
The task: a password is valid if it has at least 8 characters, contains both uppercase and lowercase letters, contains at least one digit, and contains at least one special character.

The plan in plain language: *look at every character and note which kinds we have seen. At the end, check that we have seen every kind and that the password is long enough.* Each "have we seen it?" is a **flag**, as in Notebook 03.

````python
def validate_password(password):
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False

    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True
        elif char.isdigit():
            has_digit = True
        else:
            has_special = True       # anything that is not a letter or a digit

    problems = []
    if len(password) < 8:
        problems.append("at least 8 characters")
    if not has_upper:
        problems.append("an uppercase letter")
    if not has_lower:
        problems.append("a lowercase letter")
    if not has_digit:
        problems.append("a digit")
    if not has_special:
        problems.append("a special character")
    return problems

# Testing with several cases, including the edge cases
for candidate in ["Python123!", "python", "PYTHON123!", "Py1!"]:
    problems = validate_password(candidate)
    if problems:
        print(f"{candidate:<12} invalid, missing: {', '.join(problems)}")
    else:
        print(f"{candidate:<12} valid")
````

Notice two design choices. The function **returns** a list of problems rather than printing, so the caller decides how to present them. And the tests at the bottom were chosen deliberately: one valid password, and one that fails each rule.

### A worked example: the Collatz sequence
Take any positive whole number. If it is even, halve it. If it is odd, multiply it by 3 and add 1. Repeat. The **Collatz conjecture** says that you will always reach 1, whatever number you start from. Nobody has ever proved it, and nobody has found a counterexample.

We do not know in advance how many steps are needed, so this is a job for `while`.

````python
def collatz_steps(n):
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        steps += 1
    return steps

for start in [6, 7, 27, 97]:
    print(f"Starting from {start}: {collatz_steps(start)} steps to reach 1")
````

> 🧠 This tiny loop holds a lesson for the rest of the course: the number of steps a program takes is not always obvious from reading it. Starting from 27 takes 111 steps, starting from 32 takes only 5.

## 5. A complete menu-driven program

The structure below appears in countless real programs: show a menu, read a choice, act on it, repeat until the user leaves. The three mini projects of Module B are all built on it.

````python
balance = 100

while True:
    print("\n1. Deposit   2. Withdraw   3. Show balance   4. Exit")
    choice = input("Choose an option: ")

    if choice == "1":
        amount = float(input("Amount to deposit: "))
        balance += amount
        print(f"Deposited {amount}.")
    elif choice == "2":
        amount = float(input("Amount to withdraw: "))
        if amount > balance:
            print("Insufficient funds.")
        else:
            balance -= amount
            print(f"Withdrew {amount}.")
    elif choice == "3":
        print(f"Your balance is {balance}.")
    elif choice == "4":
        print("Goodbye!")
        break
    else:
        print("Invalid option, please choose 1 to 4.")
````
