## 1. Asking yes/no questions

### Intuition first
Until now our programs ran every line, always, in the same order. Real programs must *react*: show a warning only if the password is wrong, apply a discount only if the customer qualifies. Every such reaction begins with a question whose answer is **yes or no**.

### Formal definition
A **comparison** is an expression that produces a `bool`: either `True` or `False`.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `5 > 8` | `False` |
| `<` | less than | `5 < 8` | `True` |
| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `5 <= 4` | `False` |

````python
age = 20

print(age >= 18)      # is the person an adult?
print(age == 30)      # is the person exactly 30?
print(age != 30)      # is the person NOT 30?

number = 12
print(number % 2 == 0)   # is the number even?
````

> ⚠️ **Common pitfalls**
>
> - `=` and `==` are different. One `=` **stores** a value (`age = 20`). Two `==` **ask** whether two values are equal (`age == 20`). Mixing them up is the most common beginner mistake in every programming language.

## 2. The `if` statement

### Intuition first
An `if` statement is a gate. If the condition is `True`, the gate opens and the code inside runs. If it is `False`, Python skips that code and carries on below it.

### Anatomy
```python
if condition:
    statement_1      # these lines run only when the condition is True
    statement_2
next_statement       # this line always runs
```
Two details are essential: the **colon** `:` at the end of the `if` line, and the **indentation** of the lines that belong to it.

````python
temperature = 34

if temperature > 30:
    print("It is hot today.")
    print("Remember to drink water.")

print("This line always runs.")
````

Change `temperature` to `20` and run the cell again. The two indented lines are skipped, and the last line still runs.

### Indentation defines the block
In many languages indentation is just decoration. In Python it is **part of the grammar**: the indented lines are the ones that belong to the `if`. The standard is **four spaces**, and the notebook adds them for you after a colon.

## 3. Two paths: `if` / `else`

`else` supplies the alternative: the code that runs when the condition is `False`. Exactly one of the two blocks runs, never both, never neither.

````python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
````

````python
# A second example: are you old enough to vote?
age = int(input("Enter your age: "))

if age >= 18:
    print("You are old enough to vote.")
else:
    years_left = 18 - age
    print(f"You can vote in {years_left} years.")
````

## 4. Many paths: `elif`

### Intuition first
Often there are more than two possibilities. `elif` (short for *else if*) adds further conditions. Python checks them **from top to bottom and stops at the first one that is true**. The final `else` catches everything that is left.

````python
number = float(input("Enter a number: "))

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")
````

### A worked example: a grading scale
Because Python stops at the first true condition, each `elif` can rely on the fact that all conditions above it were false. When we reach `grade >= 80`, we already know the grade is below 90, so there is no need to write `grade >= 80 and grade < 90`.

````python
grade = int(input("Enter your grade (0-100): "))

if grade >= 90:
    letter = "A"
elif grade >= 80:
    letter = "B"
elif grade >= 70:
    letter = "C"
elif grade >= 60:
    letter = "D"
else:
    letter = "F"

print(f"Your letter grade is {letter}.")
````

> ⚠️ **Common pitfalls**
>
> - **Order matters.** If the first test were `grade >= 60`, then a grade of 95 would stop there and receive a `D`. Always test the most demanding condition first.
> - `elif` versus separate `if`s: a chain of `elif` runs **at most one** block. Several separate `if` statements are checked independently, so **several** of them may run. Try it in the next cell.

````python
grade = 95

print("With separate ifs:")
if grade >= 60:
    print("  at least a D")
if grade >= 70:
    print("  at least a C")
if grade >= 80:
    print("  at least a B")

print("With an elif chain:")
if grade >= 80:
    print("  at least a B")
elif grade >= 70:
    print("  at least a C")
elif grade >= 60:
    print("  at least a D")
````

## 5. Combining conditions: `and`, `or`, `not`

### Intuition first
Real rules often have more than one part: *"you may enter if you have a ticket **and** you are over 18"*, *"the shop is closed on Saturday **or** Sunday"*.

### Formal definition

| Operator | The result is `True` when... | Example | Result |
|---|---|---|---|
| `and` | **both** sides are true | `5 > 3 and 2 > 1` | `True` |
| `or` | **at least one** side is true | `5 > 3 or 2 > 9` | `True` |
| `not` | the condition is false (it flips the answer) | `not 5 > 3` | `False` |

````python
# Is the number between 10 and 20 (inclusive)?
number = int(input("Enter a number: "))

if number >= 10 and number <= 20:
    print("The number is between 10 and 20.")
else:
    print("The number is outside the range.")
````

> 🧠 Python lets you write a range check the way mathematicians do: `10 <= number <= 20` means exactly the same as `number >= 10 and number <= 20`, and it is easier to read.

````python
day = "Sunday"
is_holiday = False

if day == "Saturday" or day == "Sunday":
    print("It is the weekend.")

if not is_holiday:
    print("It is not a public holiday.")
````

### A worked example: divisible by 3, by 5, or by both
The most specific case (both) must be tested **first**. If we tested "divisible by 3" first, the number 15 would stop there and we would never learn that it is divisible by 5 as well.

````python
number = int(input("Enter a number: "))

if number % 3 == 0 and number % 5 == 0:
    print("Divisible by both 3 and 5.")
elif number % 3 == 0:
    print("Divisible by 3 only.")
elif number % 5 == 0:
    print("Divisible by 5 only.")
else:
    print("Divisible by neither 3 nor 5.")
````

## 6. Decisions inside decisions

An `if` block may contain another `if`. This is called **nesting**, and each level is indented four more spaces. Use it when a second question only makes sense after the first has been answered.

````python
balance = 1000
amount = float(input("How much would you like to withdraw? "))

if amount > 0:
    if amount <= balance:
        balance = balance - amount
        print(f"Please take your cash. New balance: {balance}")
    else:
        print("Insufficient funds.")
        retry = float(input("Enter a smaller amount: "))
        if retry <= balance:
            balance = balance - retry
            print(f"Please take your cash. New balance: {balance}")
else:
    print("The amount must be positive.")
````

> 🧠 Deep nesting is hard to read. If you find yourself three or four levels deep, check whether `and` or `elif` can flatten the structure.

## 🔑 Key takeaways

- A **comparison** (`==`, `!=`, `<`, `>`, `<=`, `>=`) produces `True` or `False`. Remember: `=` stores, `==` compares.
- `if` runs a block only when its condition is true; `else` supplies the alternative; `elif` adds more cases.
- **Indentation** is how Python knows which lines belong to a block.
- In an `if` / `elif` chain, Python stops at the **first** true condition, so order matters: test the most specific case first.
- `and` needs both sides true, `or` needs at least one, `not` flips the answer.
- Separate `if` statements are checked independently; an `elif` chain runs at most one block.
