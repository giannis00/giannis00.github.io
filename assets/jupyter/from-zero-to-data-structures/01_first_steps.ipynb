## 1. Your first program

### Intuition first
A **program** is a list of instructions for the computer, written in a language it understands. Python is one such language, and it is a friendly one: its instructions read almost like English. The computer follows your instructions **one line at a time, from top to bottom**, and does exactly what they say, no more and no less.

### How this notebook works
This page is made of **cells**. Text cells (like this one) explain things. Code cells contain Python. To run a code cell, click on it and press **`Shift + Enter`**, or click the play button on its left. The result appears directly below the cell.

Try it now with the traditional first program of every programmer:

````python
# Anything after a "#" symbol is a COMMENT.
# Python ignores comments. They are notes for the humans reading the code.

print("Hello, World!")
````

### The anatomy of the statement
- `print` is a **function**: a named action that Python already knows how to perform.

- The quotation marks `" "` mark the beginning and end of a piece of **text**.

`print()` can show several things at once if you separate them with commas. Python puts a space between them for you.

````python
print("Python is fun")
print("My favourite number is", 7)
print("Two plus three is", 2 + 3)
print()                      # an empty print() produces a blank line
print("Done.")
````

## 2. Variables: giving names to values

### Intuition first
A **variable** is a labelled box. You put a value in the box, write a name on the label, and from then on you can use the name instead of the value.

### Formal definition
The statement `name = value` is called an **assignment**. The `=` sign does **not** mean "is equal to" as in mathematics. It means *"take the value on the right and store it under the name on the left."*

### Why it matters
Without variables a program could only ever use the values typed directly into it. With variables it can remember things, update them, and work with data it has never seen before.

````python
name = "John"
age = 25

print(name)
print(age)
print("My name is", name, "and I am", age, "years old.")
````

### A variable can change
The box can be emptied and refilled. A very common pattern is to compute a new value **from the old one**: Python first works out the right-hand side, then stores the result.

````python
x = 25
print("x starts as", x)

x = x + 5          # take the current value of x, add 5, store it back in x
print("after adding 5:", x)

x += 10            # a shorthand that means exactly:  x = x + 10
print("after adding 10 more:", x)
````

### Naming rules and conventions
A name may contain letters, digits and underscores, and it may not start with a digit. Beyond the rules, good programmers follow a convention: names are written in **lowercase with underscores**, and they **describe what the value means**.

| Good | Poor | Why |
|---|---|---|
| `total_price` | `tp` | A reader should not have to guess |
| `student_age` | `x` | The name documents the code |
| `first_name` | `FirstName` | Lowercase with underscores is the Python style |

> 🧠 You will read your own code far more often than you write it. A clear name is a small gift to your future self.

## 3. Core data types

### Intuition first
Every value has a **type**, which tells Python what kind of thing it is and what can be done with it. You can add two numbers, but adding a number to a word makes no sense, and the type is how Python knows the difference.

### The four types you will use constantly

| Type | Name | Examples | Used for |
|---|---|---|---|
| `int` | integer | `10`, `-3`, `0` | whole numbers |
| `float` | floating point | `10.5`, `3.14`, `-0.1` | numbers with a decimal point |
| `str` | string | `"Hello"`, `"42"` | text |
| `bool` | boolean | `True`, `False` | yes/no answers |

The built-in function `type()` tells you the type of any value.

````python
print(type(10))
print(type(10.5))
print(type("Hello, World!"))
print(type(True))
````

````python
# The same works on variables: the type belongs to the value inside the box
favourite_movie = "Inception"
age = 25
height = 1.75
is_student = True

print(favourite_movie, "->", type(favourite_movie))
print(age, "->", type(age))
print(height, "->", type(height))
print(is_student, "->", type(is_student))
````

### Converting between types
Sometimes a value has the wrong type for what you want to do. The functions `int()`, `float()` and `str()` convert a value into that type. This is called **casting**.

````python
text_number = "42"            # this is TEXT that happens to look like a number
print(type(text_number))

real_number = int(text_number)   # now it is a real integer
print(real_number + 1)

print(float("3.5") * 2)       # text -> float
print(str(100) + " percent")  # number -> text, so it can be glued to other text
print(int(7.9))               # float -> int simply cuts off the decimals: 7
````

> ⚠️ **Common pitfalls**
>
> - `"42"` and `42` are different things. The first is text, the second is a number. `"42" + 1` raises a `TypeError`.
> - `int(7.9)` gives `7`, not `8`. Casting to `int` cuts, it does not round. Use `round(7.9)` to round.

## 4. Arithmetic

Python is an excellent calculator. Four of the operators are the ones you know from school, and three are new and surprisingly useful.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | addition | `15 + 4` | `19` |
| `-` | subtraction | `15 - 4` | `11` |
| `*` | multiplication | `15 * 4` | `60` |
| `/` | division (always gives a float) | `15 / 4` | `3.75` |
| `//` | whole division (drops the remainder) | `15 // 4` | `3` |
| `%` | modulo (the remainder) | `15 % 4` | `3` |
| `**` | power | `2 ** 3` | `8` |

````python
x = 15
y = 4

print("Addition:      ", x + y)
print("Subtraction:   ", x - y)
print("Multiplication:", x * y)
print("Division:      ", x / y)
print("Whole division:", x // y)
print("Remainder:     ", x % y)
print("Power:         ", x ** 2)
````

### Operator precedence
Python follows the same order as mathematics: powers first, then multiplication and division, then addition and subtraction. Use brackets whenever you want a different order, or simply to make your intention clear.

````python
print(2 + 3 * 4)       # multiplication first -> 14
print((2 + 3) * 4)     # brackets first       -> 20

a = 12
b = 8
c = 5
print((a + b) * c / 2)   # (12 + 8) * 5 / 2 = 50.0
````

### The modulo operator is more useful than it looks
`%` gives the remainder of a division. Two classic uses:

- A number is **even** exactly when `number % 2` is `0`.
- Splitting a total into bigger and smaller units, such as minutes into hours and minutes.

````python
print(10 % 2)    # 0 -> 10 is even
print(7 % 2)     # 1 -> 7 is odd

# A worked example: convert 135 minutes into hours and minutes
total_minutes = 135
hours = total_minutes // 60      # how many whole hours fit
minutes = total_minutes % 60     # what is left over
print(total_minutes, "minutes is", hours, "hours and", minutes, "minutes")
````

## 5. Working with text

A **string** is a sequence of characters between quotation marks. Single quotes `'...'` and double quotes `"..."` are both fine, as long as you close with the same kind you opened with.

Two operators work on strings in a way that may surprise you:

- `+` **joins** two strings together (this is called *concatenation*).
- `*` **repeats** a string a number of times.

````python
first_name = "Giannis"
last_name = "Tsioulis"

full_name = first_name + " " + last_name    # note the space in the middle
print("Hello, my name is " + full_name + "!")

print("ha" * 3)
print("-" * 30)          # a quick way to draw a line
print(len(full_name))    # len() counts the characters, including the space
````

### f-strings: the modern, readable way
Gluing strings with `+` becomes clumsy quickly, and it fails as soon as a number is involved. An **f-string** solves both problems. Put the letter `f` before the opening quote, and write any variable or calculation inside curly braces `{ }`. Python replaces each pair of braces with its value.

````python
name = "John"
age = 30
height = 1.756

print(f"My name is {name} and I am {age} years old.")
print(f"Next year I will be {age + 1}.")
print(f"My height is {height:.2f} m.")    # :.2f means "show 2 decimal places"
````

> ⚠️ **Common pitfalls**
>
> - Forgetting the `f`: `print("I am {age}")` prints the braces literally.
> - `"Age: " + 30` raises a `TypeError`, because text and numbers cannot be joined with `+`. Use an f-string, or `str(30)`.

## 6. Reading input from the user

So far every value was typed into the code. The `input()` function makes a program **interactive**: it shows a message, pauses, and waits for the user to type an answer.

**Important:** `input()` *always* returns a **string**, even when the user types digits. To calculate with the answer, cast it with `int()` or `float()` first.

> Run the cell below and type your answer in the box that appears under it, then press `Enter`.

````python
name = input("What is your name? ")
print(f"Hello, {name}! Welcome to Python.")
````

Now an example that reads a **number**. Notice the `int(...)` wrapped around `input(...)`. Without it the value would be text and the arithmetic would fail.

````python
age = int(input("How old are you? "))
print(f"In ten years you will be {age + 10}.")
````

## 7. Putting it all together

Let us combine everything from this notebook into one small, complete program: a shop that works out a bill. Read each line and connect it to the section it came from.

````python
# A small program using input, casting, arithmetic and f-strings
product = input("What are you buying? ")
price = float(input("Price of one item (euros): "))
quantity = int(input("How many? "))

total = price * quantity

print("-" * 30)
print(f"Product:  {product}")
print(f"Quantity: {quantity}")
print(f"Total:    {total:.2f} euros")
print("-" * 30)
````

## 🔑 Key takeaways

- A program runs **one line at a time, top to bottom**, and does exactly what it says.
- `print()` displays output; comments start with `#` and are ignored by Python.
- **Variables** store values with `=`; their values can change, and `+=` is a handy shorthand.
- The four core types are `int`, `float`, `str` and `bool`; inspect them with `type()`, convert them with `int()`, `float()`, `str()`.
- Besides `+ - * /`, remember `//` (whole division), `%` (remainder) and `**` (power).
- **f-strings** (`f"...{value}..."`) are the clean way to mix text and values.
- `input()` reads from the user and **always returns a string**: cast it before doing maths.
