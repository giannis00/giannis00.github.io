## 1. Modules: using code that others wrote

### Intuition first
Python comes with a large collection of ready-made tools, organised into **modules**. A module is a toolbox: you open the one you need and use what is inside. To open a toolbox, you **import** it.

### How importing works
After `import random`, every function of that module is available through the module name and a dot.

````python
import random

print(random.randint(1, 6))      # a random whole number from 1 to 6, BOTH included
print(random.randint(1, 6))
print(random.randint(1, 6))
````

Run the cell several times. The numbers change on every run: we have just simulated three throws of a dice.

### The three functions we need

| Function | What it gives |
|---|---|
| `random.randint(a, b)` | a random integer between `a` and `b`, both included |
| `random.choice(sequence)` | one random element of a list or string |
| `random.random()` | a random float between 0 and 1 |

````python
choices = ["rock", "paper", "scissors"]
print(random.choice(choices))

coin = random.choice(["heads", "tails"])
print("The coin shows", coin)
````

> ⚠️ **Common pitfalls**
>
> - `random.randint(1, 6)` **includes** 6. This is different from `range(1, 6)`, which stops at 5.
> - Never name your own file or variable `random`: it would hide the module.

### A quick experiment: is the dice fair?
Before building games, let us use a loop and a dictionary (the counting pattern from Notebook 05) to throw a dice 6,000 times. A fair dice should show each face roughly 1,000 times.

````python
counts = {}

for throw in range(6000):
    face = random.randint(1, 6)
    counts[face] = counts.get(face, 0) + 1

for face in range(1, 7):
    print(f"Face {face}: {counts[face]} times")
````

## 2. Project 1: Guess the Number

**The game.** The computer picks a secret number between 1 and 20. The player guesses. After each guess the computer says *too low* or *too high*, until the player finds the number.

### Stage 1: one guess
We always start with the smallest version that does something.

````python
secret_number = random.randint(1, 20)

guess = int(input("Guess the number (1-20): "))

if guess == secret_number:
    print("Correct!")
else:
    print(f"Wrong. The number was {secret_number}.")
````

### Stage 2: keep guessing, with hints
We do not know how many guesses the player will need, so we use a `while` loop. The hints turn a game of pure luck into a game of logic.

````python
secret_number = random.randint(1, 20)
guess = 0

while guess != secret_number:
    guess = int(input("Guess the number (1-20): "))
    if guess < secret_number:
        print("Too low.")
    elif guess > secret_number:
        print("Too high.")

print("Correct! Well done.")
````

### Stage 3: count the attempts and package it as a function
Two improvements. A counter records the number of attempts, and the whole game becomes a function, so that it can be played again with one line and with any range of numbers.

````python
def play_guessing_game(low, high):
    secret_number = random.randint(low, high)
    attempts = 0

    while True:
        guess = int(input(f"Guess the number ({low}-{high}): "))
        attempts += 1

        if guess < secret_number:
            print("Too low.")
        elif guess > secret_number:
            print("Too high.")
        else:
            print(f"Correct! You needed {attempts} attempts.")
            return attempts

play_guessing_game(1, 20)
````

> 🧠 **A question to carry with you.** What is the smartest way to guess? If you always guess the middle of the remaining range, every answer throws away **half** of the possibilities. For 1 to 20 you never need more than 5 guesses, and for 1 to 1,000,000 you never need more than 20. This strategy has a name, *binary search*, and Notebook 13 is devoted to it.

## 3. Project 2: Rock–Paper–Scissors

**The game.** The player and the computer each choose rock, paper or scissors. Rock beats scissors, scissors beats paper, paper beats rock.

### The design
Before writing any code, we split the game into small jobs. Each job becomes one function:

| Function | Its single job |
|---|---|
| `get_computer_choice()` | return a random choice |
| `get_user_choice()` | ask the player until the answer is valid |
| `decide_winner(user, computer)` | return who won |

Then a short main part connects them. This is the method of Notebook 06 applied to a whole program: **break it down, solve each piece, test each piece.**

### Stage 1: the computer chooses

````python
def get_computer_choice():
    choices = ["rock", "paper", "scissors"]
    return random.choice(choices)

# Test the function on its own before going further
for i in range(5):
    print(get_computer_choice(), end="  ")
````

### Stage 2: deciding the winner
There are nine possible combinations, but we do not need nine cases. Either the two choices are equal (a tie), or the player has one of the three winning combinations, or, in every remaining case, the computer wins.

````python
def decide_winner(user, computer):
    if user == computer:
        return "tie"
    elif (user == "rock" and computer == "scissors") or \
         (user == "scissors" and computer == "paper") or \
         (user == "paper" and computer == "rock"):
        return "user"
    else:
        return "computer"

# Because the function has no randomness and no input, it is easy to test
print(decide_winner("rock", "scissors"))    # user
print(decide_winner("rock", "paper"))       # computer
print(decide_winner("paper", "paper"))      # tie
````

The backslash `\` at the end of a line tells Python that the statement continues on the next line. It is used here only to keep a long condition readable.

> 🧠 Notice that `decide_winner` returns a plain word and does **not** print. That makes it easy to test, and it lets the caller decide what to do with the result: show a message, update a score, or both.

### Stage 3: reading a valid choice from the player

````python
def get_user_choice():
    while True:
        choice = input("Enter rock, paper, or scissors: ").lower()
        if choice in ["rock", "paper", "scissors"]:
            return choice
        print("Invalid choice, try again.")

print("You chose:", get_user_choice())
````

### Stage 4: one complete round

````python
user_choice = get_user_choice()
computer_choice = get_computer_choice()

print(f"You chose {user_choice}, computer chose {computer_choice}.")

winner = decide_winner(user_choice, computer_choice)
if winner == "tie":
    print("It's a tie!")
elif winner == "user":
    print("You win!")
else:
    print("Computer wins!")
````

### Stage 5: many rounds, with a score
The final version wraps a round in a game loop and keeps two counters. Because every piece already works, this stage is short.

````python
def play_rock_paper_scissors():
    user_score = 0
    computer_score = 0

    while True:
        user_choice = get_user_choice()
        computer_choice = get_computer_choice()
        print(f"You chose {user_choice}, computer chose {computer_choice}.")

        winner = decide_winner(user_choice, computer_choice)
        if winner == "user":
            user_score += 1
            print("You win this round!")
        elif winner == "computer":
            computer_score += 1
            print("Computer wins this round!")
        else:
            print("It's a tie!")

        print(f"Score: you {user_score} - {computer_score} computer")

        again = input("Play again? (y/n): ").lower()
        if again != "y":
            break

    print("Thanks for playing!")

play_rock_paper_scissors()
````

## 4. What made this manageable?

Look back at how the two games were built:

- We started from the **smallest working version** and added one feature at a time.
- Each function had **one job** and could be tested on its own.
- Functions that compute **returned** their result instead of printing it.
- The main loop only *connected* pieces that already worked.

This is how all software is built, whatever its size.

## 🔑 Key takeaways

- A **module** is a toolbox of ready-made functions; `import` opens it.
- `random.randint(a, b)` includes both ends; `random.choice(seq)` picks one element.
- Build programs **in stages**: the smallest working version first, then one feature at a time.
- Give each function **one job**, and test it on its own.
- Functions that decide something should **return** the answer and leave the printing to the caller.
- A **game loop** is `while True` plus a question that lets the player leave.
