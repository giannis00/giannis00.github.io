## 1. What is a function?

### Intuition first
You have been using functions since your first line of code: `print()`, `input()`, `len()`, `type()`. Each is a **named piece of code that does one job**. You do not need to know how `len()` works inside. You give it something, and it gives you back an answer.

Now you will write your own. Think of a function as a **machine**: values go in, the machine does its work, a result comes out.

### Why it matters
- **Reuse.** Write the code once, use it as many times as you like.
- **Readability.** `calculate_average(grades)` says what happens. Ten lines of arithmetic do not.
- **Fixing in one place.** If the logic changes, you change it in the function and every use is updated.
- **Thinking in pieces.** A big problem becomes several small ones, each solved by one function.

## 2. Defining and calling a function

### Anatomy of a function
```python
def function_name(parameter1, parameter2):
    block                 # the body: what the function does
    return result         # the value handed back to the caller
```
- `def` announces a new function.
- The **parameters** are the names of the inputs.
- The indented **body** is the code that runs.
- `return` sends a result back.

**Defining** a function only teaches Python what it means. Nothing runs until you **call** it, by writing its name followed by brackets.

````python
def greet():
    print("Hello!")
    print("Welcome to the course.")

# Nothing has been printed yet. Now we call the function, twice:
greet()
greet()
````

### Parameters: passing information in
A function becomes far more useful when it can work with different values. The names inside the brackets of the definition are **parameters**. The actual values supplied in the call are **arguments**.

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Anna")       # the argument "Anna" is copied into the parameter name
greet("Kostas")
````

## 3. Returning values with `return`

### Intuition first
The functions above *show* something on the screen, but they hand nothing back. Most functions should instead **compute a value and return it**, so that the rest of the program can store it, print it, or use it in further calculations.

````python
def add_numbers(num1, num2):
    return num1 + num2

result = add_numbers(3, 4)        # the call is replaced by the returned value, 7
print(result)

print(add_numbers(10, 5) * 2)     # a returned value can be used in an expression
````

### Printing is not returning
This distinction confuses nearly everyone at first, so let us look at it directly.

- `print` shows a value **to a human**. The program itself cannot use it afterwards.
- `return` hands a value **to the program**. Nothing appears on the screen unless you print it.

A function without a `return` statement returns the special value `None`, which means "nothing".

````python
def add_and_print(a, b):
    print(a + b)            # shows the result, returns nothing

def add_and_return(a, b):
    return a + b            # returns the result, shows nothing

x = add_and_print(2, 3)     # prints 5
y = add_and_return(2, 3)    # prints nothing

print("x is", x)            # None: there was nothing to store
print("y is", y)            # 5
````

> 🧠 A good rule: a function should **return** its result and let the caller decide whether to print it. Functions that return are reusable. Functions that only print are not.

### `return` ends the function
As soon as Python reaches a `return`, the function stops and hands back the value. This lets us write decisions inside a function very cleanly.

````python
def find_max(num1, num2):
    if num1 > num2:
        return num1
    return num2        # reached only if the first return did not happen

print(find_max(8, 3))
print(find_max(2, 9))
````

### Functions that return `True` or `False`
A function that answers a yes/no question returns a boolean. Notice how short it can be: the comparison itself already *is* the answer.

````python
def is_even(number):
    return number % 2 == 0

print(is_even(10))
print(is_even(7))

# The result can be used directly in an if statement
if is_even(42):
    print("42 is even.")
````

### A worked example: converting minutes
A function can return **more than one value**, separated by commas. The caller receives them in the same order.

````python
def convert_minutes(minutes):
    hours = minutes // 60
    remaining = minutes % 60
    return hours, remaining

h, m = convert_minutes(135)
print(f"135 minutes is {h} hours and {m} minutes.")
````

## 4. Default parameter values

A parameter can be given a **default value** in the definition. If the caller does not supply that argument, the default is used.

````python
def power(base, exponent=2):
    return base ** exponent

print(power(5))        # exponent not given -> uses the default 2 -> 25
print(power(5, 3))     # exponent given     -> 125

def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Anna"))
print(greet("Anna", "Good morning"))
````

> ⚠️ **Common pitfalls**
>
> - Parameters with defaults must come **after** those without: `def f(a=1, b)` is a `SyntaxError`.
> - Calling a function with the wrong number of arguments, for example `add_numbers(3)`, raises a `TypeError`.

## 5. Variable scope

### Intuition first
A function is like a room with a one-way window. Variables created **inside** the function live only in that room: they are called **local**, and they disappear when the function finishes. The rest of the program cannot see them.

This is a feature, not a limitation. It means a function cannot accidentally damage a variable somewhere else just because it happens to use the same name.

````python
def calculate_total(price, quantity):
    total = price * quantity      # 'total' is LOCAL: it exists only in here
    return total

total = 0                         # a different variable, outside the function
result = calculate_total(5, 3)

print("Returned by the function:", result)
print("The outer 'total' is unchanged:", total)
````

> ⚠️ **Common pitfalls**
>
> - Trying to use a local variable outside its function raises a `NameError`. If the rest of the program needs a value, `return` it.

## 6. Functions that work on lists and strings

A parameter can be anything, including a list or a string. Inside, we combine the function with the loops and decisions of the previous notebooks. This is the shape of nearly every algorithm you will meet later.

````python
def average(numbers):
    total = 0
    for num in numbers:
        total += num
    return total / len(numbers)

print(average([1, 2, 3, 4]))
print(average([10, 20]))
````

````python
def count_vowels(s):
    count = 0
    for char in s.lower():        # .lower() makes the test case-insensitive
        if char in "aeiou":
            count += 1
    return count

print(count_vowels("hello"))
print(count_vowels("PROGRAMMING"))
````

### Functions calling functions
Once a function exists, other functions can use it. Large programs are built exactly like this: small, trusted pieces combined into bigger ones.

````python
def is_even(number):
    return number % 2 == 0

def count_even(numbers):
    count = 0
    for n in numbers:
        if is_even(n):            # reuse the function we already have
            count += 1
    return count

print(count_even([1, 2, 3, 4, 5, 6]))
````

    cleaned = ""
    for char in s.lower():
        if char != " ":

````python
# Your solution here:
````

```python
def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, n):
        if n % divisor == 0:
            return False      # found a divisor: stop immediately
    return True

for n in range(30):
    if is_prime(n):
        print(n, end=" ")
```

## 🔑 Key takeaways

- A **function** is a named, reusable block of code: define it with `def`, run it by **calling** it.
- **Parameters** receive the arguments passed in the call; `return` hands a result back and ends the function.
- **Printing is not returning.** A function with no `return` gives back `None`.
- **Default values** make parameters optional.
- Variables created inside a function are **local** and invisible outside it.
- Functions can take lists and strings, and can call other functions. Programs are built from small functions combined.
