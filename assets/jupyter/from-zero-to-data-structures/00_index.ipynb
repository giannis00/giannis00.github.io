Welcome. This course takes you from your very first line of Python all the way to building your own stacks, linked lists, hash tables and search trees, and to reasoning about how fast a program runs before you even run it. Every notebook opens **directly in your browser** with Google Colab: there is nothing to install and nothing to download.

Each concept is first explained in plain words, then shown in small worked examples, and then practised, a lot. That last step is deliberate: programming is a skill, and skills are built by doing. Every notebook ends with a generous set of exercises, each with a hidden solution.

## How to use these notebooks

- Work through them **in order**, each one builds on the last.
- **Run every code cell yourself** (click it and press `Shift + Enter`). Then change something and run it again.
- Attempt every exercise **before** opening the solution. Being stuck for a few minutes is where the learning happens.
- When a cell asks you to type something, look for the small input box that appears under the cell.
- Use **File → Save a copy in Drive** to keep your own editable version of each notebook.

## A note on the learning philosophy

There is a single thread running through the whole course, one simple question: **"how many steps does this take?"** You will first meet it quietly, when a loop runs ten times. It returns when a loop sits inside another loop, when we time code with a stopwatch, when we compare two sorting methods, and finally when we choose between a list, a linked list, a hash table and a tree. Data structures and algorithms are, in the end, different answers to that one question.

We also never take a claim on faith. Whenever we say that one approach is faster than another, we **measure it** and plot the result.
