**Instructor:** Ioannis Tsioulis

Welcome. This course is about one of the oldest ambitions in computing: getting a machine to work with human language. It starts with a single string of characters and ends with systems that search, translate, summarise and answer questions. Every notebook opens **directly in your browser** with Google Colab: there is nothing to install and nothing to download.

Each idea is first explained in plain words, then built **by hand** in a few lines of Python, and only then handed over to a professional library. That order is deliberate. A tokeniser, a language model or a word embedding stops being a black box the moment you have written a small one yourself.

## Who this course is for

You should be comfortable with Python (variables, loops, functions, lists and dictionaries). From Module D onwards a little NumPy helps, and Module E assumes you have met neural networks and PyTorch before.

This course does **not** re-teach those foundations. They are covered in the companion course *From Python to Deep Learning & Fine-Tuning*, and wherever a notebook here relies on an idea from there, it says so and points to the exact notebook. Everything in this course is about **language**.

## How to use these notebooks

- Work through them **in order**, each one builds on the last.
- **Run every code cell yourself** (click it and press `Shift + Enter`). Then change the text it works on and run it again. Language is full of surprises, and the best way to find them is to try your own sentences.
- Attempt every exercise **before** opening the solution.
- Some notebooks download a small corpus or a pretrained model on first use. This takes a few seconds in Colab.
- Use **File → Save a copy in Drive** to keep your own editable version of each notebook.

## A note on the learning philosophy

There is a single question running through the whole course: **"how do we turn text into numbers without losing what it means?"** A computer can only calculate, so every method in natural language processing is, in the end, an answer to that question. The first answer is the simplest one possible, counting words. It is surprisingly useful and obviously incomplete. Each module after that keeps what worked in the previous answer and repairs what did not: structure, context, similarity, order, and finally meaning.

We also never take a claim on faith. When we say that one method is better than another, we **measure it**.
